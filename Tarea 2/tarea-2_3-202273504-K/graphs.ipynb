```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Cargar los datos desde el archivo CSV
file_path = 'results.csv'  # Actualiza con la ruta correcta si es necesario
df = pd.read_csv(file_path)

# Filtrar datos que no tienen DNF en la columna 'time_ns'
df = df[df['time_ns'] != 'DNF']

# Convertir la columna 'time_ns' a numérica
df['time_ns'] = pd.to_numeric(df['time_ns'])
df['length'] = pd.to_numeric(df['length'])
df['minimum_cost'] = pd.to_numeric(df['minimum_cost'])

# Mostrar los primeros registros para confirmar la carga
print(df.head())
```

```python
## Gráfico que compara los tiempos del algoritmo de fuerza bruta con los distintos datasets
plt.figure(figsize=(14, 8))
for dataset in df['dataset'].unique():
    subset = df[(df['algorithm'] == 'brute force') & (df['dataset'] == dataset)]
    plt.plot(subset['length'], subset['time_ns'], marker='o', label=f'Brute Force - {dataset}')

plt.title('Comparación de Tiempos de Ejecución por Longitud (Algoritmo de Fuerza Bruta)')
plt.xlabel('Longitud de la secuencia')
plt.ylabel('Tiempo de ejecución (ns)')
plt.xticks(range(1, df['length'].max() + 1, 1))  # Eje x con incrementos de 1
plt.yscale('log')
plt.yticks([10**i for i in range(int(np.log10(df['time_ns'].min())), int(np.log10(df['time_ns'].max())) + 1)])  # Ticks en potencias de 10
plt.legend()
plt.grid(True, which='both', linestyle='--')
plt.show()
```

```python
## Gráfico que compara los tiempos del algoritmo de fuerza bruta con los distintos datasets
plt.figure(figsize=(14, 8))
for dataset in df['dataset'].unique():
    subset = df[(df['algorithm'] == 'dynamic programming') & (df['dataset'] == dataset)]
    plt.plot(subset['length'], subset['time_ns'], marker='o', label=f'Dynamic Programming - {dataset}')

plt.title('Comparación de Tiempos de Ejecución por Longitud (Algoritmo de Programación Dinámica)')
plt.xlabel('Longitud de la secuencia')
plt.ylabel('Tiempo de ejecución (ns)')
plt.xticks(range(1, df['length'].max() + 1, 1))  # Eje x con incrementos de 1
plt.legend()
plt.grid(True, which='both', linestyle='--')
plt.show()
```

```python
# Gráfico que compara los tiempos en el dataset 'normal' entre fuerza bruta y programación dinámica
plt.figure(figsize=(14, 8))
algorithms_to_compare = ['brute force', 'dynamic programming']
for algorithm in algorithms_to_compare:
    subset = df[(df['dataset'] == 'normal') & (df['algorithm'] == algorithm)]
    plt.plot(subset['length'], subset['time_ns'], marker='o', label=f'{algorithm.capitalize()} - normal')

plt.title('Comparación de Tiempos de Ejecución por Longitud (Dataset Normal)')
plt.xlabel('Longitud de la secuencia')
plt.ylabel('Tiempo de ejecución (ns)')
plt.xticks(range(1, df['length'].max() + 1, 1))  # Eje x con incrementos de 1
plt.yscale('log')  # Escala logarítmica en el eje y
plt.yticks([10**i for i in range(int(np.log10(df['time_ns'].min())), int(np.log10(df['time_ns'].max())) + 1)])  # Ticks en potencias de 10
plt.legend()
plt.grid(True, which='both', linestyle='--')
plt.show()
```

```python
# Gráfico que compara los tiempos en el dataset 'normal' entre fuerza bruta y programación dinámica
plt.figure(figsize=(14, 8))
algorithms_to_compare = ['brute force', 'dynamic programming']
for algorithm in algorithms_to_compare:
    subset = df[(df['dataset'] == 'empty') & (df['algorithm'] == algorithm)]
    plt.plot(subset['length'], subset['time_ns'], marker='o', label=f'{algorithm.capitalize()} - empty')

plt.title('Comparación de Tiempos de Ejecución por Longitud (Dataset Empty)')
plt.xlabel('Longitud de la secuencia')
plt.ylabel('Tiempo de ejecución (ns)')
plt.xticks(range(1, df['length'].max() + 1, 1))  # Eje x con incrementos de 1
plt.legend()
plt.grid(True, which='both', linestyle='--')
plt.show()
```
